Testing the ERFNet model on anomaly segmentation

```python
import subprocess
from itertools import product

methods = ["msp", "maxlogit", "entropy"]
anomaly_datasets = {
    "RoadAnomaly" : "./anomaly/Anomaly_Datasets/Validation_Dataset/RoadAnomaly/images/*.jpg",
    "RoadAnomaly21" : "./anomaly/Anomaly_Datasets/Validation_Dataset/RoadAnomaly21/images/*.png",
    "RoadObstacle21" : "./anomaly/Anomaly_Datasets/Validation_Dataset/RoadObsticle21/images/*.webp",
    "FS_static" : "./anomaly/Anomaly_Datasets/Validation_Dataset/fs_static/images/*.jpg",
    "FS_LostFound" : "./anomaly/Anomaly_Datasets/Validation_Dataset/FS_LostFound_full/images/*.png"
}
for method_name, dataset_name in product(methods, anomaly_datasets.keys()) :
    print(f"Dataset {dataset_name} method {method_name}")
    args = ["python", "./erfnet_anomaly.py", "--input", anomaly_datasets[dataset_name], "--method", method_name]
    output = subprocess.run(args, capture_output = True, text = True)
    print(output.stdout, flush = True)
```

Test the EoMT models on anomaly segmentation 

```python
import subprocess
from itertools import product

methods = ["msp", 
           "maxlogit", 
           "entropy", 
           "rba", 
           "temperature"]
anomaly_datasets = {
    "RoadAnomaly" : "./anomaly/Anomaly_Datasets/Validation_Dataset/RoadAnomaly/images/*.jpg",
    "RoadAnomaly21" : "./anomaly/Anomaly_Datasets/Validation_Dataset/RoadAnomaly21/images/*.png",
    "RoadObstacle21" : "./anomaly/Anomaly_Datasets/Validation_Dataset/RoadObsticle21/images/*.webp",
    "FS_static" : "./anomaly/Anomaly_Datasets/Validation_Dataset/fs_static/images/*.jpg",
    "FS_LostFound" : "./anomaly/Anomaly_Datasets/Validation_Dataset/FS_LostFound_full/images/*.png"
}
models = {
    "Cityscapes trained" : {
        "weights" : "./checkpoints/pretrained/eomt_cityscapes.bin",
        "config" : "./configs/eomt_semantic_trained_on_cityscapes.yaml"
    },
    "COCO trained" : {
        "weights" : "./checkpoints/pretrained/eomt_coco.bin",
        "config" : "./configs/eomt_semantic_trained_on_coco.yaml"
    },
    "COCO finetuned" : {
        "weights" : "./checkpoints/finetuned/coco_finetuned_on_cityscapes.bin",
        "config" : "./configs/eomt_finetune_model_on_cityscapes.yaml"
    }
}
for method_name, dataset_name, model in product(methods, anomaly_datasets.keys(), models.keys()) :
    print(f"Dataset {dataset_name} method {method_name} model {model}")
    args = ["python", "./eomt_anomaly.py", 
            "--configPath", models[model]["config"],
            "--loadWeights", models[model]["weights"],
            "--input", anomaly_datasets[dataset_name],
            "--method", method_name
            ]
    output = subprocess.run(args, capture_output = True, text = True)
    print(output.stdout, flush = True)
```
